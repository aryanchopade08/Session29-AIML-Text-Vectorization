```python
import pandas as pd

df = pd.read_csv("cleaned_emotions.csv")

df.head()
```

```python
from sklearn.model_selection import train_test_split

X = df["text_no_stopwords"]
y = df["emotions"]

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42
)

print("X_train:", X_train.shape)
print("X_test:", X_test.shape)
print("y_train:", y_train.shape)
print("y_test:", y_test.shape)
```

Q2

```python
from sklearn.feature_extraction.text import CountVectorizer

vectorizer = CountVectorizer()

X_train_bow = vectorizer.fit_transform(X_train)
X_test_bow = vectorizer.transform(X_test)

print("X_train_bow shape:", X_train_bow.shape)
print("X_test_bow shape:", X_test_bow.shape)
```

```python
from sklearn.feature_extraction.text import CountVectorizer

vectorizer = CountVectorizer()

X_train_bow = vectorizer.fit_transform(X_train)
X_test_bow = vectorizer.transform(X_test)

print("X_train_bow shape:", X_train_bow.shape)
print("X_test_bow shape:", X_test_bow.shape)
```

```python
print("First 20 vocabulary words:")
print(vectorizer.get_feature_names_out()[:20])
```

Q3


```python
from sklearn.naive_bayes import MultinomialNB

model_bow = MultinomialNB()

model_bow.fit(X_train_bow, y_train)

y_pred_bow = model_bow.predict(X_test_bow)

from sklearn.metrics import accuracy_score

accuracy_bow = accuracy_score(y_test, y_pred_bow)

print("Bag of Words Accuracy:", accuracy_bow)
```

Q5

```python
from sklearn.feature_extraction.text import CountVectorizer

bigram_vectorizer = CountVectorizer(ngram_range=(1, 2))

X_train_bigram = bigram_vectorizer.fit_transform(X_train)
X_test_bigram = bigram_vectorizer.transform(X_test)

print("X_train_bigram shape:", X_train_bigram.shape)
print("X_test_bigram shape:", X_test_bigram.shape)

print("Total bigram features:", len(bigram_vectorizer.get_feature_names_out()))

print("Sum of bigram features:", X_train_bigram.sum())
```

Q6

```python
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score

model_bigram = MultinomialNB()

model_bigram.fit(X_train_bigram, y_train)

y_pred_bigram = model_bigram.predict(X_test_bigram)

accuracy_bigram = accuracy_score(y_test, y_pred_bigram)

print("N-gram Bag of Words Accuracy:", accuracy_bigram)
print("Basic BoW Accuracy:", accuracy_bow)
```

Q7

```python
from sklearn.feature_extraction.text import TfidfVectorizer

tfidf = TfidfVectorizer()

X_train_tfidf = tfidf.fit_transform(X_train)
X_test_tfidf = tfidf.transform(X_test)

print("X_train_tfidf shape:", X_train_tfidf.shape)
print("X_test_tfidf shape:", X_test_tfidf.shape)
```

```python
model_tfidf = MultinomialNB()

model_tfidf.fit(X_train_tfidf, y_train)

y_pred_tfidf = model_tfidf.predict(X_test_tfidf)

accuracy_tfidf = accuracy_score(y_test, y_pred_tfidf)

print("TF-IDF Accuracy:", accuracy_tfidf)
```

```python
print("Basic BoW Accuracy :", accuracy_bow)
print("N-gram BoW Accuracy:", accuracy_bigram)
print("TF-IDF Accuracy    :", accuracy_tfidf)
```

Q8

```python
model_tfidf = MultinomialNB()

model_tfidf.fit(X_train_tfidf, y_train)

y_pred_tfidf = model_tfidf.predict(X_test_tfidf)

accuracy_tfidf = accuracy_score(y_test, y_pred_tfidf)

print("TF-IDF Accuracy:", accuracy_tfidf)
```

Q9

```python
print("Comparison of Text Vectorization Methods")
print("---------------------------------------")

print("Bag of Words (Unigrams):", accuracy_bow)
print("Bag of Words (Unigrams + Bigrams):", accuracy_bigram)
print("TF-IDF:", accuracy_tfidf)

best_method = max(
    {
        "Bag of Words (Unigrams)": accuracy_bow,
        "Bag of Words (Unigrams + Bigrams)": accuracy_bigram,
        "TF-IDF": accuracy_tfidf
    },
    key={
        "Bag of Words (Unigrams)": accuracy_bow,
        "Bag of Words (Unigrams + Bigrams)": accuracy_bigram,
        "TF-IDF": accuracy_tfidf
    }.get
)

print("\nBest Method:", best_method)
```
